```python
import random

# Mapping values directly to their custom weight assignments
WHEEL_POOL = {
    "100": 15.0,
    "150": 12.0,
    "200": 10.0,
    "250": 9.0,
    "300": 8.0,
    "400": 7.0,
    "500": 6.0,
    "750": 5.0,
    "1000": 4.0,
    "1500": 3.0,
    "2000": 2.0,
    "2500": 1.5,
    "3000": 1.0,
    "4000": 0.8,
    "5000": 0.5,
    "7500": 0.2,
    "10000": 0.1,
    "DOUBLE_POINTS": 4.0,
    "MULTIPLIER_BONUS": 2.0,
    "FREE_SPINS": 4.0,
    "NO_POINTS": 4.9,
}

SEGMENTS = list(WHEEL_POOL.keys())
WEIGHTS = list(WHEEL_POOL.values())


def spin_21_wheel():
    return random.choices(SEGMENTS, weights=WEIGHTS, k=1)[0]


def run_session(starting_spins=1):
    total_points = 0
    spins_left = starting_spins

    current_multiplier = 1
    double_flag = False

    while spins_left > 0:
        spins_left -= 1
        landing = spin_21_wheel()

        if landing.isdigit():
            gained = int(landing) * current_multiplier
            if double_flag:
                gained *= 2
            total_points += gained
            current_multiplier, double_flag = 1, False  # Consume modifiers

        elif landing == "DOUBLE_POINTS":
            spins_left += 1
            double_flag = True

        elif landing == "MULTIPLIER_BONUS":
            current_multiplier = 5

        elif landing == "FREE_SPINS":
            spins_left += 2

        elif landing == "NO_POINTS":
            current_multiplier, double_flag = 1, False  # Clear active modifiers

    return total_points


if __name__ == "__main__":
    SIMULATIONS = 100000
    payout_accumulator = sum(run_session() for _ in range(SIMULATIONS))

    print(f"Validated across {SIMULATIONS:,} standard games...")
    print(
        f"📊 Balanced Expected Return: {payout_accumulator / SIMULATIONS:.2f} points per session."
    )
```
