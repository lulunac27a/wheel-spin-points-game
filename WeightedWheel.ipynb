```python
import random

# Mapping values directly to their custom weight assignments
WHEEL_POOL = {  # points list
    "100": 15.0,
    "150": 12.0,
    "200": 10.0,
    "250": 9.0,
    "300": 8.0,
    "400": 7.0,
    "500": 6.0,
    "750": 5.0,
    "1000": 4.0,
    "1500": 3.0,
    "2000": 2.0,
    "2500": 1.5,
    "3000": 1.0,
    "4000": 0.8,
    "5000": 0.5,
    "7500": 0.2,
    "10000": 0.1,
    "DOUBLE_POINTS": 4.0,
    "MULTIPLIER_BONUS": 2.0,
    "FREE_SPINS": 4.0,
    "NO_POINTS": 4.9,
}

SEGMENTS = list(WHEEL_POOL.keys())  # segments of the wheel
WEIGHTS = list(WHEEL_POOL.values())  # weights of the segments


def spin_21_wheel():  # simulates a single spin of the wheel and returns the result based on the defined weights
    return random.choices(SEGMENTS, weights=WEIGHTS, k=1)[0]


def run_session(
    starting_spins=1,
):  # runs a single session of the game, simulating spins and calculating total points based on the outcomes
    total_points = 0
    spins_left = starting_spins

    current_multiplier = 1
    double_flag = False

    while (
        spins_left > 0
    ):  # continues spinning until there are no spins left, applying the effects of any special segments landed on
        spins_left -= 1
        landing = spin_21_wheel()

        if (
            landing.isdigit()
        ):  # if the landing segment is a point value, calculate the points gained based on the current multiplier and any active double points effect
            gained = int(landing) * current_multiplier
            if (
                double_flag
            ):  # if the double points effect is active, double the points gained
                gained *= 2
            total_points += gained
            current_multiplier, double_flag = 1, False  # Consume modifiers

        elif (
            landing == "DOUBLE_POINTS"
        ):  # if the landing segment is DOUBLE_POINTS, increment the spins left and set the double_flag to True to indicate that the next points gained will be doubled
            spins_left += 1
            double_flag = True

        elif (
            landing == "MULTIPLIER_BONUS"
        ):  # if the landing segment is MULTIPLIER_BONUS, increment the spins left and set the current_multiplier to 5 to indicate that the next points gained will be multiplied by 5
            current_multiplier = 5

        elif (
            landing == "FREE_SPINS"
        ):  # if the landing segment is FREE_SPINS, increment the spins left by 2 to give the player additional spins
            spins_left += 2

        elif (
            landing == "NO_POINTS"
        ):  # if the landing segment is NO_POINTS, simply continue to the next iteration without modifying total_points or spins_left
            current_multiplier, double_flag = 1, False  # Clear active modifiers

    return total_points  # return the total points accumulated during the session


if __name__ == "__main__":
    SIMULATIONS = 100000
    payout_accumulator = sum(
        run_session() for _ in range(SIMULATIONS)
    )  # run multiple sessions of the game and accumulate the total points gained across all sessions to calculate the average expected return per session

    print(f"Validated across {SIMULATIONS:,} standard games...")
    print(
        f"📊 Balanced Expected Return: {payout_accumulator / SIMULATIONS:.2f} points per session."
    )
```
