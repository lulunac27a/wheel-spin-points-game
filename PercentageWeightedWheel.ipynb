```python
import random
import time


class PercentageWeightedWheel:
    def __init__(self, starting_points=100, starting_spins=5, verbose=True):
        self.points = starting_points
        self.highest_points = starting_points
        self.free_spins = starting_spins
        self.verbose = verbose
        self.total_spins_cast = 0

        self.wheel_outcomes = ["POINTS", "PERCENTAGE", "FREE_SPINS", "DOUBLE"]
        self.wheel_weights = [0.55, 0.20, 0.15, 0.10]

        self.points_list = [
            100,
            150,
            200,
            250,
            300,
            400,
            500,
            750,
            1000,
            1500,
            2000,
            2500,
            3000,
            4000,
            5000,
            7500,
            10000,
        ]
        self.points_weights = [
            0.20,
            0.16,
            0.14,
            0.12,
            0.10,
            0.08,
            0.06,
            0.04,
            0.03,
            0.02,
            0.015,
            0.01,
            0.008,
            0.006,
            0.005,
            0.004,
            0.002,
        ]

        self.percentage_list = [0.10, 0.20, 0.25, 0.50, 0.75, 1.00]
        self.percentage_weights = [0.34, 0.26, 0.18, 0.12, 0.07, 0.03]
        self.percentage_signs = [1, -1]
        self.percentage_sign_weights = [0.5, 0.5]

        self.spins_payout_list = [2, 3, 5]
        self.spins_payout_weights = [0.50, 0.30, 0.20]

    def roll_discrete_points(self):
        return random.choices(self.points_list, weights=self.points_weights, k=1)[0]

    def roll_percentage_change(self):
        magnitude = random.choices(
            self.percentage_list, weights=self.percentage_weights, k=1
        )[0]
        sign = random.choices(
            self.percentage_signs, weights=self.percentage_sign_weights, k=1
        )[0]
        return magnitude * sign

    def roll_nested_free_spins(self):
        return random.choices(
            self.spins_payout_list, weights=self.spins_payout_weights, k=1
        )[0]

    def spin(self):
        if self.free_spins <= 0:
            if self.verbose:
                print("\nOut of spins! Game Over.")
            return False

        self.free_spins -= 1
        self.total_spins_cast += 1
        landed_outcome = random.choices(
            self.wheel_outcomes, weights=self.wheel_weights, k=1
        )[0]

        if landed_outcome == "POINTS":
            points_won = self.roll_discrete_points()
            self.points += points_won
            message = f"Standard Points: +{points_won:,}"
        elif landed_outcome == "PERCENTAGE":
            percentage_change = self.roll_percentage_change()
            old_points = self.points
            self.points = max(0, round(self.points * (1 + percentage_change)))
            label = f"{percentage_change:+.0%}"
            message = (
                f"Percentage Change: {label} " f"({old_points:,} -> {self.points:,})"
            )
        elif landed_outcome == "FREE_SPINS":
            spins_won = self.roll_nested_free_spins()
            self.free_spins += spins_won
            message = f"Free Spins: +{spins_won}"
        else:
            old_points = self.points
            self.points *= 2
            message = f"Double Points: {old_points:,} -> {self.points:,}"

        self.highest_points = max(self.highest_points, self.points)
        if self.verbose:
            print(message)
            self.display_hud()
        return True

    def display_hud(self):
        print("-" * 70)
        print(
            f"POINTS TOTAL: {self.points:,}  | "
            f"HIGHEST: {self.highest_points:,}  | "
            f"REMAINING SPINS: {self.free_spins}"
        )
        print("-" * 70)


game = PercentageWeightedWheel(starting_points=100, starting_spins=5, verbose=True)
game.display_hud()
while game.spin():
    pass
print(
    f"Session Terminated: {game.points:,} points | "
    f"Highest Points: {game.highest_points:,} across "
    f"{game.total_spins_cast} casts."
)
```

```python
for _ in range(100):
    wheel = PercentageWeightedWheel(
        starting_points=100, starting_spins=5, verbose=False
    )
    while wheel.spin():
        pass
    print(
        f"Percentage Game over -> Points: {wheel.points:,} | "
        f"Highest Points: {wheel.highest_points:,} | "
        f"Total spins cast: {wheel.total_spins_cast}"
    )

    zero_points = PercentageWeightedWheel(
        starting_points=0, starting_spins=10, verbose=False
    )
    while zero_points.spin():
        pass
    print(
        f"Zero Points Percentage Game over -> Points: {zero_points.points:,} | "
        f"Highest Points: {zero_points.highest_points:,} | "
        f"Total spins cast: {zero_points.total_spins_cast}"
    )
```
